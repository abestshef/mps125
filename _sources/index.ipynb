# Introduction to the module
 
Welcome to the webpages and learning materials for MPS125.

Below are some key details about the module:
```{admonition} Lecture notes

The lecture notes are provided online, and you are free to view them on a laptop/tablet or (if you have good eyesight) a phone. You are welcome to print them out but we do not supply printed notes for you. These notes detail all the mathematical knowledge you need for this semester of MPS125. Ideally you would have a copy - either printed or online - with you when at the lectures as I will refer to them. The original version of these notes were written by my colleague Jonathan Potts - many thanks to him for his hard work and sharing his resources with me.
```
```{admonition} Blackboard

This is the online repository for all information about the module. Please check this regularly (you will also get email updates).
```
```{admonition} Obtaining help with the module

Please use the Dicussion Board on Blackboard to ask questions in the first instance - I will check this regularly. You are also strongly encouraged to come along to my student drop-in hours for in-person help. These will be advertisd on Blackboard.
```
```{admonition} Homework and practice questions

At the end of each lecture there are some homework exercises. You will cover some of these in the tutorials, but the rest you should do for homework. Some of these I ask you to hand in so you can obtain feedback on your work: you can see which questions on the ‘Assignment sheet’ on Blackboard. These do not count towards your final grade (called 'formative' in university-speak).
```
```{admonition} Tutorials

There will be small-group tutorials where you will be able to obtain help on the practice questions. These are a key part of this course, as they are the main opportunity for you to sit and work on some modelling problems, with support from a tutor.
```
```{admonition} Assessment

20% of your mark for the whole module will be on a project to be handed in in January. Details are at the end of tehse lecture notes and on Blackboard, and I will talk about this in detail later in the semester. The remaining 80% will be assessed via an invigilated exam at the end of the year.
```
```{admonition} Lectures

You are encouraged to engage in **active learning**. I will encourage you to complete bits of working, discuss with your neighbour and vote on various aspects during lectures. I hope for there to be a welcoming and inclusive environment in our lectures. If you feel you are not fully welcomed in the classes for any reason, please think about coming for a chat during an office hour.
```
