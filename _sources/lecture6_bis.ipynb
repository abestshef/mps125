# Lecture 7. An application: water pollution in the Peak District

> *'I thought she’d be back in three weeks and we’d go wandering the Peaks'*

Last lecture, we worked through two key ways to solve first-order linear ODEs: separation of variables and integrating factors. We applied these to two fairly straight-forward examples. Now let's apply these methods to a more interesting real-world scenario.

Among the many delights of the Peak District National Park (have you been yet? You should - take a bus or a train) are the three connected reservoirs of Howden, Derwent and Ladybower. Suppose a company decides to illegally dump pollution into Howden reservoir. How will this pollution spread downstream? 

Let us define a few terms first. We will use the letters $H$, $D$ and $L$ to denote the amount of pollution in the three respective reservoirs. We will assume the pollution quickly becomes well-mixed into the body of water. The flow rate of water (and, since it is well-mixed, therefore also the flow rate of pollution) from Howden into Derwent is given by $\phi$, and the flow-rate from Derwent into Ladybower as $\rho$. The initial amount of pollution is $P_0$. For simplicity we will assume there is no out-flow from Ladybower (perhaps after the pollution event immediate measures were taken to stop the flow and limit the spread). 

We can then write out these three equations as,

$$
\begin{array}{ll}
\frac{\mathrm{d} H}{\mathrm{~d} t} & = -\phi H(t) \\
\frac{\mathrm{d} D}{\mathrm{~d} t} & = \phi H(t)-\rho D(t) \\ 
\frac{\mathrm{d} L}{\mathrm{~d} t} & = \rho D(t)
\end{array}
$$

This gives us a set of three *simultaneous* differential equations to solve. You will look into how to deal with such situations more fundamentally in another module. However, in this case we can effectively solve them one at a time.

### 7.1 The dynamics in Howden

Note that the first equation depends only on the variable $H(t)$ on both sides of the equality. We can therefore treat this case entirely on its own. More helpfully still, since we have no constant term, we can solve this through separation of variables.


````{prf:example}
Solve

$$
\begin{equation*}
\frac{\mathrm{d} H}{\mathrm{~d} t}=-\phi H
\end{equation*}
$$

**Solution.**
```{toggle}
Separate the variables to get 

$$
\begin{equation*}
\frac{\mathrm{d} H}{H}=-\phi dt 
\end{equation*}
$$

Then we can integrate both sides to get,

$$
\begin{equation*}
\ln(H)=-\phi t+C_1
\end{equation*}
$$

Finally we take the exponential of both sides, leaving,

$$
\begin{align*}
H(t)&=e^{-\phi t+C_1} \\
&=e^{C_1}e^{-\phi t}\\
&=P_0e^{-\phi t}
\end{align*}
$$
````

We've got a (reasonably) nice expression for the pollution level in Derwent now, but it still contains a constant $C_2$. Do we know an initial condition we can use to replace this with something more meaningful? Think back: the pollution was initially dumped into Howden reservoir, meaning that at our start point ($t=0$) there was no pollution in Derwent. So we can take,

$$
\begin{align*}
D(0)=0&=\frac{\phi}{\rho-\phi}P_0e^{0}+C_2e^{0} \\
C_2&=-\frac{\phi}{\rho-\phi}P_0
\end{align*}
$$

This means our full solution is,

$$
\begin{equation*}
D(t)=\frac{\phi}{\rho-\phi}P_0\left[e^{-\phi t}-e^{-\rho t}\right].
\end{equation*}
$$

```{note}
If the two flow rates are equal, $\rho=\phi$ then our solution doesn't make sense. A solution does exist, we just have to restart our workings replacing $\rho$ with $\phi$. This is left as a homework exercise for you.
```

### 7.3 The dynamics in Ladybower

Finally we come to the equation for pollution in Ladybower,

$$
\begin{equation*}
\frac{\mathrm{d} L}{\mathrm{~d} t}=\rho D.
\end{equation*}
$$

Again, initially we might worry that it is the 'wrong' dependent variable on the RHS, until we remember we now have a full solution for this, so the equation becomes, 

$$
\begin{equation*}
\frac{\mathrm{d} L}{\mathrm{~d} t}= \frac{\rho\phi}{\rho-\phi}P_0\left[e^{-\phi t}-e^{-\rho t}\right].
\end{equation*}
$$

Now there are no dependent variables at all in the RHS! This means it is just a standard integration. 

$$
\begin{align*}
L(t)=\int\frac{\mathrm{d} L}{\mathrm{~d} t}dt&= \int\frac{\rho\phi}{\rho-\phi}P_0\left[e^{-\phi t}-e^{-\rho t}\right]dt. \\
&=\frac{\rho\phi}{\rho-\phi}P_0\left[\frac{e^{-\phi t}}{-\phi}-\frac{e^{-\rho t}}{-\rho}\right]+C_3\\
&=\frac{-\left[{e^{-\phi t}}\rho-{e^{-\rho t}}\phi\right]}{\rho-\phi}P_0+C_3\\
&=\frac{\left[{e^{-\rho t}}\phi-{e^{-\phi t}}\rho\right]}{\rho-\phi}P_0+C_3
\end{align*}
$$

Finally, we should try and replace the constant $C_3$ with something more meaningful. Again, at the start of the event we know $L(0)=0$, so we can say,

$$
\begin{align*}
L(0)=0&= \frac{\left[{e^{0}}\phi-{e^{0}}\rho\right]}{\rho-\phi}P_0+C_3\\
C_3&=-\frac{\left[\phi-\rho\right]}{\rho-\phi}P_0\\
C_3&=P_0
\end{align*}
$$

So we have the final equation for pollution in Ladybower as,

$$
\begin{equation*}
L(t)= \left[\frac{{e^{-\rho t}}\phi-{e^{-\phi t}}\rho}{\rho-\phi}+1\right]P_0.
\end{equation*}
$$

Well that was nice. We had three first-order linear ordinary differential equations that we have solved by three different methods: seperation of variables, an integrating factor, and plain-old integration.

### 7.4: Answering real-world questions with a model

We might now feel very satisfied with ourselves for having completed all these calculations. However, our *mathematical modelling* work has not yet finished. The workers from the Environment Agency or Peak District National Park don't just want to be given some equations, as nice as they may look. They will have some practical questions for us. We will consider one possible query here. To make things even more realistic, let's take some reasonable values for some of our quantities. We will say the amount of pollution dumped is 0.5 megalitres (Ml). The bulk flow rate between the reservoirs is 150 Ml/day. The volume of Howden is around 7,500 Ml and Derwent is around 30,000 Ml, giving respective flow rates of $\phi=0.02$ and $\rho=0.005$.


#### "When will pollution in Derwent reservoir peak?"

We have an expression for the amount of pollution in Derwent, which we can substitute our values into:

$$
\begin{align*}
D(t)&=\frac{\phi}{\rho-\phi}P_0\left[e^{-\phi t}-e^{-\rho t}\right]\\
&=-\frac{2}{3}\left[e^{-0.02 t}-e^{-0.005 t}\right].
\end{align*}
$$

How can we find the peak of this? Imagine what the graph of this will look like. We know it starts at 0, it's going to increase as the pollution flows in, and then decrease again as it flows out. So we expect there to be a maximum to this function, and that will occur where $dD(t)/dt=0$.

$$
\begin{align*}
\frac{dD(t)}{dt}=0&=-\frac{2}{3}\left[-0.02e^{-0.02 t}+0.005e^{-0.005 t}\right]\\
&=-0.02e^{-0.02 t}+0.005e^{-0.005 t}\\
4e^{-0.02 t}&=e^{-0.005 t}.
\end{align*}
$$

Taking logs of both sides we get,

$$
\begin{align*}
\ln(4)-0.02 t&=-0.005 t\\
0.015t&=\ln(4)\\
t&\approx92\textrm{ days}.
\end{align*}
$$

So, if no further action were taken, we'd have to wait 3 months for the pollution to peak in Derwent, suggesting this will be a long-term issue.
