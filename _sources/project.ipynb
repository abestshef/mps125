# MPS125 Project: Saving red squirrels

**This assessment is worth 20\% of your grade for MPS125.  You should hand it in on Blackboard, via the 'Assessment Information / Written Project' folder, by 2pm Thursday 28th January.**

The UK's native red squirrel population has had a precipitous decline since the introduction of the non-native grey squirrel in the late 1800s, due to a mix of competition and disease. One potential conservation tool being considered is encouraging populations of pine martens, another native species. Pine martens are predators of both red and grey squirrels, but are particularly succesful at hunting the grey squirrels: in regions where the three species coexist naturally, grey squirrel populations have been found to be much smaller, leading to more resilient red squirrel populations.

For this project you will take the role of a mathematical modelling consultant, such as might work at BIOSS (https://www.bioss.ac.uk/) or CEH (https://www.ceh.ac.uk/). You have been contacted by a local Wildlife Trust, who are looking into the pros and cons of introducing pine martens into the region to help limit the grey squirrel population. Your task is to produce some initial modelling of the predator-prey dynamics of pine martens and grey squirrels, analyse the results and draw some conclusions.

Let $G(t)$ and $P(t)$ be the number of grey squirels and pine martens at time $t$, respectively.  You are provided with the following initial model for their dynamics:

\begin{align}
	\label{prey}
	G(t+1)&=G(t)+rG(t)-pG(t)P(t), \\
	\label{pred}
	P(t+1)&=P(t)+pcP(t)G(t)-dP(t),
\end{align}

where $r$, $p$, $c$, and $d$ are positive constants. Based on field-studies, the current best estimates for the parameters are: $r=0.5$, $p=0.0001$, $c=0.5$, $d=0.1$. There is some uncertainty in the population numbers. Set $G(0)=4000+N$ where $N$ is the last three digits of your registration number.  Set $P(0)=4000+M$ where $M$ is the number given by the 5th, 6th, and 7th digits of your registration number.  (For example, if your registration number is 160153427 then $G(0)=4427$ and $P(0)=4534$.)

Your tasks are as follows:

1. Explain how to interpret the constants $r$, $p$, $c$, and $d$ biologically.
2.   Calculate $G(t)$ and $P(t)$ for $t<40$ and plot the resulting curves (of $G(t)$ and $P(t)$ against $t$) on a single graph.
3. Make at least one change to the model to gain wider insight into the possible behaviours. Ideas include using different parameter values or initial conditions. You are also welcome to make any changes to the model you believe are sensible for the conservation issue in question (though be careful of spending too much time on this - you are a busy consultant!).
4. Provide a biological interpretation of your results and what your advice would be based on this work.
5. Write up what you have found by doing tasks 1-4 in a single report, using no more than 2 A4 pages.  Hand this report in according to the instructions at the top of this page.

## Tips on how to write a (good) report

It is not uncommon for maths students to complain that they don’t like writing, perhaps even claiming it’s why they study maths. Unfortunately, I am here to tell you that **good writing is an essential mathematical skill**. Even those of us who have pursued academic careers - perhaps the most ‘pure’ form of mathematical profession - need to be able to write research papers, grant applications, teaching notes, etc. clearly and coherently. In professions in industry, the civil service, teaching, etc. this will be even more true. 

So, how do you write a report that covers mathematical material? There is no single answer to this, but I will give you a few pointers of things that I think are important:

**Choose a good piece of software**

Handwritten reports will occasionally be acceptable (informal minutes from a meeting, say) but you should really be using typesetting software. A few common choices are:

* LaTeX - custom-designed for writing mathematical work (most of you are learning this in Mathematical Investigation Skills). Excellent for neat formatting of your report, especially if there is significant mathematical content and/or plots. The main downside is accessibility - screen readers cannot parse the equations.

* Microsoft Word / Google Docs - classic easy-to-use ‘what you see is what you get’ word processors. Mathematical notation can be used but can be a faff, as can neat formatting of long documents. 

* Markdown - a halfway house, that mostly is a plain word processor but will typeset equations using LaTeX commands. Importantly it can be parsed by screen readers. This is what I use for all your notes (with some embellishment…). It is optimised for web browsers so doesn't always work so well for printed reports.

**Format your document nicely**

What you produce should be easy to read. Think of when you read a textbook, or a magazine article, or an online news story - how are these laid out to make them easily digestible? A few suggestions from me:
* Include a title and sensible section headers.
* Avoid big blocks of text that have no paragraph breaks. 
* Equally, avoid long chains of mathematical workings with no text to explain your workings.
* Intersperse figures and plots with the text.
* Keep to sensible font sizes and margin widths - it should all be readable at 100% scaling on a laptop.

**Think carefully about who your reader is**

A maths lecturer specialising in this area is going to have a very different baseline knowledge to a recruitment officer at a pharmaceutical company who knows nothing about maths but is handling your job application. What can you expect them to know, and what is their guiding interest? *Remember, in this case you are writing for someone from the Wildlife Trust*.

**Tell a story**

I don’t mean make it up. I mean guide us through the work from start to finish. What is the underlying question that motivates the work? Then how have you approached it? What did you find? What do we learn from this? This is the classic structure: introduction, methods, results, conclusions. Sometimes it’ll make sense to do things a bit differently, but it’s generally a good starting point. 

**Be wary of generative AI**

It may be tempting to just ask a large language model to “write me a 800-word report on a discrete-time predator-prey model” but please don’t. Would I be able to spot it? Probably, though not certainly - things are always improving but my experience is you still get warning signs. More importantly, you won’t learn anything, in terms of doing the actual modelling, in terms of writing, or in the feedback you get.

Different people, and certainly different academics, have different takes on the role of AI. For me, in this context, I see it has uses as a spelling and grammar checker, especially for those whose first language is not English. I can also see its use to get you started with some code or point you to a bit of extra reading but you should thoroughly check what it gives you is what you wanted. If you do use generative AI in any capacity at all you **must** declare it in your submission.
